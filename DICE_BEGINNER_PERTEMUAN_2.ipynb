**Genre Paling Umum:** Identifikasi genre yang paling sering muncul.

```python
# Ambil kolom 'listed_in' (genre)
# Split setiap genre yang dipisahkan koma lalu jadikan satu list panjang
genre_series = df['listed_in'].astype(str).dropna().str.split(', ') # Changed 'data' to 'df' and converted to string
genre_list = genre_series.explode()

# Hitung jumlah kemunculan tiap genre
genre_counts = genre_list.value_counts().head(10)

# Visualisasi
plt.figure(figsize=(12, 6))
sns.barplot(x=genre_counts.values, y=genre_counts.index, palette='viridis')
plt.title('10 Genre Paling Umum di Netflix')
plt.xlabel('Jumlah Tayangan')
plt.ylabel('Genre')
plt.tight_layout()
plt.show()
```

**Top 10 Tahun dengan Banyaknya Film Rilis**

Hitung banyaknya film yang rilis di tiap tahunnya.

```python
# Filter for movies only
movies_df = df[df['type'] == 'Movie']

# Group by release year and count the number of movies
movie_release_counts = movies_df.groupby('release_year')['show_id'].count().sort_values(ascending=False)

# Select the top 10 years with the most movie releases
top_10_years = movie_release_counts.head(10)

# Visualize the results using a bar plot
plt.figure(figsize=(12, 6))
sns.barplot(x=top_10_years.index, y=top_10_years.values, palette='viridis')
plt.title('Top 10 Tahun dengan Banyaknya Film yang Rilis di Netflix')
plt.xlabel('Release Year')
plt.ylabel('Number of Movies')
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels for better readability
plt.tight_layout()
plt.show()
```

**Top 10 Genre Netflix dengan rata rata durasi terpanjang**

menghitung rata rata diurasi film terpanjang di netflix.

```python
# Contoh data
data = {
    'listed_in': [
        'Action & Adventure, Anime Features, Classic Movies',
        'Action & Adventure, Anime Features, Horror Movies',
        'Action & Adventure, Anime Features, International Movies',
        'Action & Adventure, Anime Features, Sci-Fi & Fantasy',
        'Action & Adventure, Children & Family Movies',
        'Action & Adventure, Children & Family Movies, Classic Movies',
        'Action & Adventure, Children & Family Movies, Comedies'
    ],
    'duration': [120, 115, 125, 130, 90, 95, 85]  # Contoh durasi dalam menit
}

df = pd.DataFrame(data)

# Mengonversi durasi ke numerik (jika belum)
df['duration'] = pd.to_numeric(df['duration'], errors='coerce')

# Menghitung durasi rata-rata per genre
genre_duration = df.groupby('listed_in')['duration'].mean().sort_values(ascending=False).head(10)

# Membuat plot
plt.figure(figsize=(10, 6))
sns.barplot(x=genre_duration.values, y=genre_duration.index, palette='magma')

# Menambahkan judul dan label
plt.title('10 Genre dengan Durasi Rata-rata Terpanjang di Netflix', pad=20)
plt.xlabel('Durasi Rata-rata (menit)')
plt.ylabel('Genre')

# Menyesuaikan layout agar label tidak terpotong
plt.tight_layout()
plt.show()
```
